Bastin et al.

```python
import rasterio
import numpy as np

main_raster = ".../Bastin_PotC_Adj_mskd.tif"
mask_raster = ".../Bastin_PotC_Adj.tif"

with rasterio.open(main_raster) as src:
    data = src.read(1).astype(float)
    pixel_size_x = abs(src.transform.a)
    pixel_size_y = abs(src.transform.e)
    pixel_area_km2 = (pixel_size_x * pixel_size_y) / 1e6  
    nodata_main = src.nodata
    profile = src.profile  

with rasterio.open(mask_raster) as msk:
    mask_data = msk.read(1)
    nodata_mask = msk.nodata

valid_mask = (data != nodata_main) & np.isfinite(data) 
mask_condition = (mask_data > 0.1) & (mask_data != nodata_mask)

combined_mask = (data ==1) & valid_mask & mask_condition

count_valid = np.sum(combined_mask)

weighted_sum = np.sum((mask_data[combined_mask]) * pixel_area_km2)   
weighted_sum_Mha = weighted_sum / 10000  

print(f"Number of valid pixels: {count_valid}")
print(f"Unweighted area (pixel count only): {count_valid * pixel_area_km2 / 10000:.2f} Mha")
print(f"Weighted area (value × area): {weighted_sum_Mha:.2f} Mha")
print(f"Pixel size: {pixel_size_x:.2f} m × {pixel_size_y:.2f} m")
```
